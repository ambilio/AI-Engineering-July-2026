```python
arr = np.array([
    [2, 2, 2], 
    [4, 5, 6], 
    [0.1, 0.2, 0.5], 
    [3, -7, 0.3]
    ])
```

```python
arr2 = arr[:2,::2]
print("First two rows and alternate columns:\n",arr2)
```

```python
print("Add 1 to every element:\n",arr2+1)
```

```python
arr2=arr2+1
print("After adding 1:\n",arr2)
```

```python
print("Subtract 2 from every element:\n",arr2-2)
```

```python
print("Sum of array elements:", np.sum(arr2))
```
