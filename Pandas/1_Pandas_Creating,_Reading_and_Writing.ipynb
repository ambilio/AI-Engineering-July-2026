````python
import kagglehub
import os
import shutil

# Download the required datasets
datasets = {
    'pitchfork-data': 'nolanbconaway/pitchfork-data',
    'chess': 'datasnaek/chess',
    'kepler-exoplanet-search-results': 'nasa/kepler-exoplanet-search-results',
    'things-on-reddit': 'residentmario/things-on-reddit',
    'wine-reviews': 'zynicide/wine-reviews',
    'ramen-ratings': 'residentmario/ramen-ratings',
    'powerlifting-database': 'dansbecker/powerlifting-database',
    'youtube-new': 'datasnaek/youtube-new',
    '188-million-us-wildfires': 'rtatman/188-million-us-wildfires',
    'publicassistance': 'jpmiller/publicassistance'
}

# Create the base input directory
input_dir = '../input'
os.makedirs(input_dir, exist_ok=True)

print('Downloading and linking datasets...')

for folder_name, dataset_id in datasets.items():
    # Download
    download_path = kagglehub.dataset_download(dataset_id)

    # Destination path in ../input/
    dest_path = os.path.join(input_dir, folder_name)

    # Remove existing link/folder if it exists to avoid errors on re-run
    if os.path.exists(dest_path):
        if os.path.islink(dest_path):
            os.unlink(dest_path)
        else:
            shutil.rmtree(dest_path)

    # Create symbolic link
    os.symlink(download_path, dest_path)
    print(f'Linked {folder_name} to {dest_path}')

print('\nAll datasets are linked. You can now run the Setup cell below.')
````
