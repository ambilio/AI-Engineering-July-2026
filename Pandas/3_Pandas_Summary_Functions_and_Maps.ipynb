```python
import kagglehub
import os

# Download datasets
data_sources = {
    'rtatman/188-million-us-wildfires': '188-million-us-wildfires',
    'nolanbconaway/pitchfork-data': 'pitchfork-data',
    'datasnaek/chess': 'chess',
    'nasa/kepler-exoplanet-search-results': 'kepler-exoplanet-search-results',
    'dansbecker/powerlifting-database': 'powerlifting-database',
    'residentmario/ramen-ratings': 'ramen-ratings',
    'residentmario/things-on-reddit': 'things-on-reddit',
    'datasnaek/youtube-new': 'youtube-new',
    'jpmiller/publicassistance': 'publicassistance',
    'zynicide/wine-reviews': 'wine-reviews'
}

# Ensure the ../input directory exists
os.makedirs('../input', exist_ok=True)

for handle, folder_name in data_sources.items():
    path = kagglehub.dataset_download(handle)
    link_path = os.path.join('../input', folder_name)
    if not os.path.exists(link_path):
        os.symlink(path, link_path)
    print(f'Linked {folder_name} to ../input/')

print('Data source import and linking complete.')
```
